```python
import sys, time

delay = 7

def plurial(v: int)->str:
  return "s" if v>1 else ""

for i in range(delay, 0, -1):
    plur = plurial(i)
    sys.stdout.write(f"\r{i} seconde{plur} restante{plur}...")
    sys.stdout.flush()
    time.sleep(1)

print(f'\rFini d\'attendre {delay} seconde{plurial(delay)} !')
```
